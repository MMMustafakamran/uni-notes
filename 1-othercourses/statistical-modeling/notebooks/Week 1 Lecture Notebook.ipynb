# Synthetic data

How can we use the PyMC for inference and then for prediction using the Bayesian Theorm?

```python
import scipy as sp
p=0.9
trails=20

data=sp.stats.bernoulli.rvs(p=p,size=trails)
```

```python
data
```

```python
with pm.Model() as our_model:
    prior= pm.Beta('p ', alpha= 1.5 , beta= 5 )
    likelyhood= pm.Bernoulli(   'y'   ,p=prior  ,observed=data )
    t= pm.sample(100)
```
